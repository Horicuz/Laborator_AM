```python
from functools import partial
from http.server import SimpleHTTPRequestHandler, ThreadingHTTPServer
from pathlib import Path
import threading
```

```python
def _find_web_directory():
    current = Path.cwd().resolve()
    for candidate in [current, *current.parents]:
        web_dir = candidate / "web"
        if web_dir.is_dir():
            return web_dir
    raise FileNotFoundError("Could not find the web/ directory.")


class QuietHTTPRequestHandler(SimpleHTTPRequestHandler):
    def log_message(self, format, *args):
        _ = format % args
        return


WEB_DIR = _find_web_directory()
_http_server = globals().get("_http_server")


def stop_web_server():
    global _http_server

    if _http_server is None:
        print("[CLIENT 2] Web server is not running.")
        return

    _http_server.shutdown()
    _http_server.server_close()
    _http_server = None
    print("[CLIENT 2] Web server stopped.")


if _http_server is not None:
    print("[CLIENT 2] Web server already running on port 8001.")
    print("[CLIENT 2] Open this URL in your browser: http://127.0.0.1:8001")
else:
    handler = partial(QuietHTTPRequestHandler, directory=str(WEB_DIR))
    _http_server = ThreadingHTTPServer(("127.0.0.1", 8001), handler)
    threading.Thread(target=_http_server.serve_forever, daemon=True).start()
    print("[CLIENT 2] Starting the web server on port 8001...")
    print("[CLIENT 2] Open this URL in your browser: http://127.0.0.1:8001")
    print("[CLIENT 2] Run stop_web_server() to stop it.")
```
